```python
input_sentence = "After finishing my project and getting positive feedback from my team, I couldn't help but feel incredibly proud happy! 😄🎉"
```

```python
import nltk, re, string, spacy, emoji, contractions
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('punkt')
```

```python
lower_input_sentence = input_sentence.lower()
print("Lower Case: ", lower_input_sentence)
```

```python
contractions_lower_input_sentence = contractions.fix(lower_input_sentence)
print("Contractions Fixed: ", contractions_lower_input_sentence)
```

```python
demojize_contractions_lower_input_sentence = emoji.demojize(contractions_lower_input_sentence)
replace_emoji_contractions_lower_input_sentence = emoji.replace_emoji(contractions_lower_input_sentence, replace="__" )
print("Emoji Demojized: ", demojize_contractions_lower_input_sentence)
print("Emoji Replaced: ", replace_emoji_contractions_lower_input_sentence)
print("partially done")
```
